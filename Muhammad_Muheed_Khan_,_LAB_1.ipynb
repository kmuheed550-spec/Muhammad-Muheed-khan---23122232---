```python
print("===== TRAVEL AGENT =====")

destination = input("Where do you want to travel? ")
budget = int(input("Enter your budget in PKR: "))

if destination.lower() == "karachi":
    if budget >= 30000:
        print("You can visit Karachi.")
        print("Recommended: Beach, Mazar-e-Quaid and Clifton.")
    else:
        print("Your budget is low for this trip.")

if destination.lower() == "lahore":
    if budget >= 25000:
        print("You can visit Lahore.")
        print("Recommended: Badshahi Mosque, Minar-e-Pakistan and Lahore Fort.")
    else:
        print("Your budget is low for this trip.")

if destination.lower() == "islamabad":
    if budget >= 35000:
        print("You can visit Islamabad.")
        print("Recommended: Faisal Mosque, Daman-e-Koh and Monal.")
    else:
        print("Your budget is low for this trip.")

else:
    print("Sorry! This destination is not available.")
    print("Please choose Karachi, Lahore or Islamabad.")

print("Thank you for using Travel Agent!")
```
